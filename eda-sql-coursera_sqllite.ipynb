### Task 1




##### Display the names of the unique launch sites  in the space mission


```python
# Task 1: Display the names of the unique launch sites
sql_query_1 = "SELECT DISTINCT Launch_Site FROM SPACEXTABLE;"

print(sql_query_1)
result_1 = cur.execute(sql_query_1).fetchall()
# Display results
print("\nUnique Launch Sites:")
for site in result_1:
    print(site[0])
```


### Task 2


#####  Display 5 records where launch sites begin with the string 'CCA' 


```python
# Task 2: Display 5 records where launch sites begin with the string 'CCA'
sql_query_2 = "SELECT * FROM SPACEXTABLE WHERE Launch_Site LIKE 'CCA%' LIMIT 5;"

print(sql_query_2)
result_2 = cur.execute(sql_query_2).fetchall()
# Display results (Headers: FlightNumber, Date, Booster_Version, Payload_Mass__kg_, etc.)
# ... Outputting 5 full records ...
```

### Task 3




##### Display the total payload mass carried by boosters launched by NASA (CRS)


```python
# Task 3: Display the total payload mass carried by boosters launched by NASA (CRS)
sql_query_3 = "SELECT SUM(Payload_Mass__kg_) FROM SPACEXTABLE WHERE Customer = 'NASA (CRS)';"

print(sql_query_3)
result_3 = cur.execute(sql_query_3).fetchone()[0]
print(f"\nTotal Payload Mass for NASA (CRS): {result_3} kg")
```

### Task 4




##### Display average payload mass carried by booster version F9 v1.1


```python
# Task 4: Display average payload mass carried by booster version F9 v1.1
sql_query_4 = "SELECT AVG(Payload_Mass__kg_) FROM SPACEXTABLE WHERE Booster_Version = 'F9 v1.1';"

print(sql_query_4)
result_4 = cur.execute(sql_query_4).fetchone()[0]
print(f"\nAverage Payload Mass for F9 v1.1: {result_4} kg")
```

### Task 5

##### List the date when the first succesful landing outcome in ground pad was acheived.


_Hint:Use min function_ 


```python
# Task 5: List the date when the first successful landing outcome in ground pad was acheived.
sql_query_5 = "SELECT MIN(Date) FROM SPACEXTABLE WHERE Landing_Outcome = 'Success (ground pad)';"

print(sql_query_5)
result_5 = cur.execute(sql_query_5).fetchone()[0]
print(f"\nFirst successful landing date on a ground pad: {result_5}")
```

### Task 6

##### List the names of the boosters which have success in drone ship and have payload mass greater than 4000 but less than 6000


```python
# Task 6: List the names of the boosters which have success in drone ship and have payload mass greater than 4000 but less than 6000
sql_query_6 = "SELECT Booster_Version FROM SPACEXTABLE WHERE Landing_Outcome = 'Success (drone ship)' AND Payload_Mass__kg_ > 4000 AND Payload_Mass__kg_ < 6000;"

print(sql_query_6)
result_6 = cur.execute(sql_query_6).fetchall()
print("\nBooster Versions (Success Drone Ship, 4000-6000 kg):")
for booster in result_6:
    print(booster[0])
```

### Task 7




##### List the total number of successful and failure mission outcomes


```python
# Task 7: List the total number of successful and failure mission outcomes
sql_query_7 = "SELECT Mission_Outcome, COUNT(*) AS Total_Count FROM SPACEXTABLE GROUP BY Mission_Outcome;"

print(sql_query_7)
result_7 = cur.execute(sql_query_7).fetchall()
print("\nMission Outcomes Count:")
for outcome, count in result_7:
    print(f"'{outcome}': {count}")
```

### Task 8



##### List all the booster_versions that have carried the maximum payload mass, using a subquery with a suitable aggregate function.


```python
# Task 8: List all the booster_versions that have carried the maximum payload mass, using a subquery
sql_query_8 = "SELECT DISTINCT Booster_Version FROM SPACEXTABLE WHERE Payload_Mass__kg_ = (SELECT MAX(Payload_Mass__kg_) FROM SPACEXTABLE);"

print(sql_query_8)
result_8 = cur.execute(sql_query_8).fetchall()
print("\nBooster Versions with Max Payload Mass (15,600 kg):")
for booster in result_8:
    print(booster[0])
```

### Task 9


##### List the records which will display the month names, failure landing_outcomes in drone ship ,booster versions, launch_site for the months in year 2015.

**Note: SQLLite does not support monthnames. So you need to use  substr(Date, 6,2) as month to get the months and substr(Date,0,5)='2015' for year.**


```python
# Task 9: List the records for failure landing_outcomes in drone ship for months in year 2015.
# SQLLite uses SUBSTR(Date, 6, 2) as month to get the months and SUBSTR(Date, 1, 4) = '2015' for year.
sql_query_9 = "SELECT SUBSTR(Date, 6, 2) AS Month, Landing_Outcome, Booster_Version, Launch_Site FROM SPACEXTABLE WHERE Landing_Outcome = 'Failure (drone ship)' AND SUBSTR(Date, 1, 4) = '2015';"

print(sql_query_9)
result_9 = cur.execute(sql_query_9).fetchall()
print("\n2015 Failure (drone ship) Launches:")
print("{:<5} | {:<20} | {:<15} | {:<15}".format("Month", "Landing_Outcome", "Booster_Version", "Launch_Site"))
print("-" * 60)
for row in result_9:
    print("{:<5} | {:<20} | {:<15} | {:<15}".format(row[0], row[1], row[2], row[3]))
```

### Task 10




##### Rank the count of landing outcomes (such as Failure (drone ship) or Success (ground pad)) between the date 2010-06-04 and 2017-03-20, in descending order.


```python
# Task 10: Rank the count of landing outcomes between the date 2010-06-04 and 2017-03-20, in descending order.
sql_query_10 = "SELECT Landing_Outcome, COUNT(*) AS Count FROM SPACEXTABLE WHERE Date BETWEEN '2010-06-04' AND '2017-03-20' GROUP BY Landing_Outcome ORDER BY Count DESC;"

print(sql_query_10)
result_10 = cur.execute(sql_query_10).fetchall()
print("\nLanding Outcomes Count (2010-06-04 to 2017-03-20):")
print("{:<20} | {:<5}".format("Landing_Outcome", "Count"))
print("-" * 30)
for outcome, count in result_10:
    print("{:<20} | {:<5}".format(outcome, count))
```
